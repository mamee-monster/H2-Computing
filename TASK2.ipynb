```python
# TASK 2.1

# where filename is name of text file
def task2_1(filename):
    list_of_int = []
    with open(filename, "r") as file:
        read_file = file.read()
        list_of_str_num = read_file.split("\n")
        
        for num in list_of_str_num:
            list_of_int.append(int(num))
        
        return list_of_int
    
```

```python
result = task2_1('TEN.txt')
print(result)
print(len(result))
```

```python
# TASK 2.2

def task2_2(list_of_integers):
    # insertion sort
    for i in range(1, len(list_of_integers)):
        while list_of_integers[i-1] > list_of_integers[i] and i>0:
            list_of_integers[i-1], list_of_integers[i] = list_of_integers[i], list_of_integers[i-1]
            i -= 1
    return list_of_integers
```

```python
task2_2(task2_1('TEN.txt'))
```

```python
# TASK 2.3

# what is recursion: calls itself? [3m]
# 1. calls itelf recursively with smaller, similar subproblems
# 2. until base case is hit
# 3. then backtrcks to solve the original problem


def task2_3(list_of_integers):
    # base case: a list of 0 or 1 items is already sorted
    if len(list_of_integers) <= 1:
        return list_of_integers
    # partition around the first element as the pivot
    pivot = list_of_integers[0]
    smaller = [value for value in list_of_integers[1:] if value < pivot]
    larger = [value for value in list_of_integers[1:] if value >= pivot]
    # recursively quicksort each partition either side of the pivot
    return task2_3(smaller) + [pivot] + task2_3(larger)
    
    # by convention choose 1st item
    
    # if qn asks for last tiem, data[-1]
    
    # last itme: len(data) //2
```
