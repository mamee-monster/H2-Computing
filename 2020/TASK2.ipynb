```python
# TASK 2.1

import random 

def task2_1(filename, quantity, maximum):
    lst = []
    for i in range(0, quantity):
        val = random.randint(0, maximum)
        lst.append(val)
        
    with open(filename, "w") as file:
        for item in lst:
            file.write(str(item) + '\n')
            
task2_1("randomnumbers.txt", 1000, 5000)
```
