```python
# TASK 3.1            # would marks still be awarded if this can only process integer quantities?
def task3_1(quantity_of_data):
    number = ""
    unit = ""
    for bit in quantity_of_data:
        # check for quantity
        if bit.isdigit() == True:
            number += bit
        else:
            unit += bit
    
    if unit == "KB":
        zeros = "000"
        print((number + zeros))        
    elif unit == "MB":
        zeros = " 000 000"
        print((number + zeros))        
    elif unit == "GB":
        zeros = " 000 000 000"
        print((number + zeros))        
    elif unit == "TB":
        zeros = " 000 000 000 000"
        print((number + zeros))
    else:
        print("invalid data")
        
task3_1("8KB")
task3_1("8CB")
task3_1("9TB")
```
