```python
import sqlite3

connection = sqlite3.connect("school.db")
# .cursor object only needed in cases of selecting data

# PersonID Table
connection.execute("DROP TABLE IF EXISTS People")
connection.execute("""
CREATE TABLE IF NOT EXISTS People(
   PersonID INTEGER PRIMARY KEY AUTOINCREMENT,
   FullName TEXT,
   DateOfBirth TEXT,
   ScreenName TEXT,
   IsAdult INTEGER)
   """)

connection.commit()
connection.close()
```

```python
# TASK 4.1 ------
from datetime import date

class Person:
    def __init__(self, full_name, date_of_birth):  # constuctor
        self.full_name = full_name
        self.date_of_birth = date_of_birth

    def is_adult(self):
        today = date.today() #todays's date eg.2000-01-01
        year_of_birth = self.date_of_birth.split('-')[0]
        age = int(today.year) - int(year_of_birth)
        #print(age)
        if age > 18:
            return True
        return False

    def screen_name(self):
        name = ""
        for i in self.full_name:
            if i != " ":
                name += i
        name += self.date_of_birth.split("-")[1] # add month
        name += self.date_of_birth.split("-")[2] # add day
        return name
```

```python
# TASK 4.2 ------
## Staff Class
class Staff(Person):
    def screen_name(self):
        return super().screen_name() + "Staff"  # super() -> call parent's method first

    def is_adult(self):
        return True

## Student Class
class Student(Person):
    def screen_name(self):
        return super().screen_name() +"Student"

    def is_adult(self):
        return False

## read function
def read_insert():
    # list to store instantiated objects:
    ppl_objects = []
    # read function
    with open("people.txt", "r") as file:
        read_file = file.readlines()

        for p in read_file:
            p = p.strip().split(",")
            name = p[0]
            dob = p[1]
            role = p[2]
            # shorter alt:
            # name, dob, role = p[0], p[1], p[2]

            # creating instance for each person
            if p[2] == "Student":
                person = Student(name, dob)
                #print(person.full_name, person.date_of_birth)
            elif p[2] == "Staff":
                person = Staff(name, dob)
            else:
                person = Person(name, dob)  #
            ppl_objects.append(person)    
            
    # inserting into DB
    import sqlite3
    connection = sqlite3.connect("school.db")
    cursor = connection.cursor()

    #cursor.execute("DROP TABLE IF EXISTS People")

    for person in ppl_objects:
        cursor.execute("INSERT INTO People(FullName, DateOfBirth, ScreenName, IsAdult) VALUES (?,?,?,?)",(person.full_name, person.date_of_birth, person.screen_name(), person.is_adult()))
             
    connection.commit()
    recs = cursor.execute("SELECT * FROM People")
    for rec in recs:
        print(rec[1])
    connection.close()
    
```
