```python
# TASK 2.1
class QueueItem:
    def __init__(self, ID, quantity):
        self.ID = ID
        self.quantity = quantity
        
    def getID(self):
        return self.ID
    
    def getQuantity(self):
        return self.quantity
    
# TASK 2.2
class Queue:
    def __init__(self):
        self.queueList = 4 * [None] # initialise 4 null objects 
        self.headPointer = -1
        self.tailPointer = -1
        self.numberItems = 0
    
    def enqueue(self, data):
        # full q
        if self.numberItems == 4:
            print("Queue is full")
            return False
            
        # usual
        self.numberItems += 1
        self.tailPointer = (self.tailPointer+1) % 4
        self.queueList[self.tailPointer] = data
        print(f"{data.getID()} has been enqueued")
        return True
    
    def dequeue(self):
        # empty q
        if self.numberItems == 0:
            print("Queue is empty, nothing to dequeue")
            return False
        
        # usual
        remove = self.queueList[self.headPointer]
        self.headPointer = (self.headPointer + 1) % 4
        self.numberItems -= 1
        print(f"{remove.getID()} has been dequeued$")
        return remove
    
    def printQueue(self):
        for i in self.queueList:
            print(f"ID: {i.getID()} \t quantity: {i.getQuantity()}")
        
```

```python
# TASK 2.3

q = Queue()  # initialise queue

q.enqueue(QueueItem('a', 3))
q.enqueue(QueueItem('b', 6))
q.enqueue(QueueItem('c', 7))
q.enqueue(QueueItem('d', 2))
q.enqueue(QueueItem('e', 4))
```

```python
# TASK 2.4
q.dequeue()
q.dequeue()
q.dequeue()
q.dequeue()
q.dequeue()
```
