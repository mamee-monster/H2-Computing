```python
# TASK 3.1
def inputPuzzle():  
    sudoku = []

    for i in range(9):
        while True:
            row = input(f"Enter row {i+1}: ")
            array = row.split(" ")
            
            # create new list if integers (previously were strings -> allows easier manipulation for later parts)
            rowList = []
            
            for num in array:
                # length check
                if len(array) != 9:
                    print("Please input 9 digits with spacings inbetween.")
                    break
                
                # range check
                if int(num) < 0 or int(num) > 9:
                    print("Please ensure that digits are within range of 0 to 9.")
                    break
                
                # type check
                if num.isdigit() == False:
                    print("Please ensure that only digits are entered.")
                    break
                    
                rowList.append(int(num))
            #print(rowList)
                    
            # all correct,
            #valid = True
                    
            # append into 2D list:
            if rowList != []:
                sudoku.append(rowList)

            #print(sudoku)
            #print(len(sudoku))
            if len(rowList) == 9:
                break
    print(sudoku)
    return sudoku
```

```python
# TASK 3.2
def check_horizontal(array):
    count = 0
    for row in array:
        #bprint(row)
        has_dupe = len(row) != len(set(tuple(row)))
        count += 1
        if has_dupe:
            return f'duplicate at row: {count}'
    return -1
```

```python
# TASK 3.3
def check_vertical(array):
    count = 0    
    for row in array:
        column = []
        for i in range (9):
            column.append(array[i][count])
        count += 1
        #print(column)
        has_dupe = len(column) != len(set(tuple(column)))
        if has_dupe:
            return f'duplucate at column: {count}'
    return -1
            
check_vertical([[7, 8, 2, 9, 1, 3, 4, 5, 6], [1, 4, 5, 8, 7, 6, 9, 2, 3], [6, 9, 3, 6, 5, 2, 7, 1, 8], [2, 7, 9, 5, 3, 1, 8, 6, 4], [5, 6, 4, 2, 9, 8, 1, 3, 7], [3, 1, 8, 7, 6, 4, 5, 9, 2], [4, 5, 1, 6, 2, 7, 3, 8, 9], [9, 2, 7, 3, 8, 5, 6, 4, 1], [8, 3, 6, 1, 4, 9, 2, 7, 5]])
```

```python
# TASK 3.4
def check():
    puzzle = inputPuzzle()
    # returns sudoku list 
    vertical_valid = check_vertical(puzzle)
    horizontal_valid = check_horizontal(puzzle)
    
    print(vertical_valid)
    print(horizontal_valid)
    
    if vertical_valid == horizontal_valid:
        return True
    return False
    
check()
```
