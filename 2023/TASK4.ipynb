```python
# TASK 4_1
# compressing data
def read_raw(file_name):
    with open(file_name, 'r') as read_file:
        file =  read_file.read()
        file = file.split('\n')
        return file # returns: ['000', '010', '011', '011'...]

def compress(colours):
    streak = 1
    result = []
    current = colours[0]
    for colour in colours[1:]:
        if current == colour:
            streak += 1
        else:
            result.append([current, streak])
            current = colour
            streak = 1
    result.append([current, streak]) # format: [colour, no. of times repeated]

    #print(result)

    compressed_list = []

    for combi in result:
        compressed_data = combi[0]+str(combi[1])
        compressed_list.append(compressed_data)

    # print(compressed_list)

    with open('compressedimage.txt', 'w') as file:
        for data in compressed_list:
            file.write(data)
```

```python
compress(read_raw('imagefile.txt'))
```

```python
# TASK 4_2
def read_compressed(file_name):
    with open(file_name, 'r') as read_file:
        file =  read_file.read()
        return file
    
def decompress(colours):
    decompressed_list = []
    i = 0
    while i < len(colours):
        colour = colours[i:i+4]
        i += 4
        decompressed_list.append(colour)

    with open("decompressedimage.txt","w") as file:
        for colour in decompressed_list:
            streak = colour[3]
            pixel = colour[:3]
            decompress = str(pixel) * int(streak)
            #print(decompress)
            file.write(decompress)
```
