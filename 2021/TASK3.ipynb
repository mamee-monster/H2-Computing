```python
# TASK 3.1
# dynamic LL

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
class LinkedList:
    def __init__(self):
        self.head = None  # initially empty list
        
    def insert(self, integer_value):
        new_node = Node(integer_value)
        new_node.next = self.head
        self.head = new_node
        
    def delete(self, integer_value):
        # case1: empty LL
        if self.head == None:
            return None
        
        # case3: data is the head of LL
        current = self.head

        if current.data == integer_value:
            deleted = current
            self.head = current.next
            return integer_value
            
        prev = None
        # case2: data may be somewhere in LL
        while current.next is not None:
            if current.data == integer_value:
                prev.next = current.next
                return integer_value
            
            prev = current
            current = current.next
        return None
                
        
    def search(self, integer_value):
        # empty list
        if self.head == None:
            return False
        
        # item at head
        current = self.head
        if current.data == integer_value:
            return True
        
        #somwhere else in list
        while current.next is not None:
            if current.data == integer_value:
                return True
            current = current.next
        
    
    def count(self):
        count = 0
        current = self.head
        while current is not None:
            count+=1
            current = current.next
        print(count)
        return count 
        
    def to_String(self):
        current = self.head
        linkedlist = []
        while current is not None:
            linkedlist.append(current.data)
            current = current.next
        print(linkedlist[::-1])  # where: self.head = 2
    
LL = LinkedList()
LL.insert(2)
LL.insert(4)
LL.insert(3)
LL.insert(5)
LL.insert(8)
LL.insert(7)

LL.delete(4)

LL.count()

LL.to_String()
```

```python
# TASK 3.3
class Queue(LinkedList):
    def enqueue(self, integer_value): 
        new_node = Node(integer_value)
        
        # empty queue
        if self.head is None:
            super().insert(integer_value)
            return 
        # insert at end
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node
        
    def dequeue(self):
        # empty queue
        if self.head is None:
            return None
        
        # usual case
        delete = self.head
        self.head = self.head.next
        delete.next = None # ★ disconnects the deleted node from queue
        return delete

    def to_String(self):
        current = self.head
        linkedlist = []
        while current is not None:
            linkedlist.append(current.data)
            current = current.next
        print(linkedlist)        
```
