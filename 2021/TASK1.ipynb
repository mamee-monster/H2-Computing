```python
# TASK 1.1

def task1_1(input_value):
    # format check
    for n in input_value:
        if n.isdigit() == True:
            pass
        else:
            return -1
        
    # length check
    length = 0
    for n in input_value:
        length += 1
    if length != 13:
        return -1
    
    # check digit
    list_of_num = []
    for n in input_value:
        list_of_num.append(n)
    #print(list_of_num)
    reverse = list_of_num[::-1]
    #print(reverse)
        
    odd_list = []
    for o in range(0, 13, 2):
        #print(o)
        odd_num = int(reverse[o]) * 2
        odd_str = str(odd_num)
        single_odd = list(odd_str)
        odd_list.extend(single_odd)
    #print(odd_list)
        
        
    even_list = []
    for e in range(1, 13, 2):
        even = int(reverse[e])
        even_list.append(even)
    #print(even_list)
        
    odd_list.extend(even_list)
    #print(odd_list)
    
    total = 1
    for i in odd_list:
        i = int(i)
        total += i
        
    #print(total//2)
    
    check_digit = 10 - (total%10)
    return check_digit
```

```python
#TASK 1.2

print(task1_1("1234567890123"))
print(task1_1("1478934758934"))
print(task1_1("1651623565626"))
print(task1_1("1982341647346"))
```

```python
# TASK 1.3
def task1_3(input_value):
    # format check:
    for n in input_value:
        if n.isdigit() == True:
            pass
        else:
            return False
    
    # length check:
    length = 0
    for n in input_value:
        length += 1
    if length != 15:
        return False

    check = task1_1(input_value[1:-1])
        
    if check != -1:
        return True
    return False
```
